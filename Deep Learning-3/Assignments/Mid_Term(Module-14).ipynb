**Part A Answer:**
Sigmoid: Sigmoid is one of the most popular activation functions used in Machine learning and deep learning, especially for binary classification.It gives an output between 0 and 1. 

ReLU: ReLU(Reactified Linear Unit) is the most widely used activation function in modern deep learning. It gives outputs 0 for negative inputs and returns the input iteself for positive inputs. It is simple, efficient, and faster than sigmoid. 

ReLU is preferred in hidden layers of deep networks, because:
1. Its simple and fast,
2. It helps to reduce vanishing gradient problem,
3. For positive inputs ist gardient is 1, so gradients can flow more easily through deep networks,

Sigmoid can cause the vanishing gradient problem during backpropagation because its gradient becomes very small for large positive or negative inputs very close to 0. So, earlier layers learn very slowly or may almost stop learning. 

**Part B Answer:**
The difference between Batch Gradient Descent, Stochastic Gradient Descent (SGD), and Mini-batch Gradient Descent are:
Batch Gradient Descent: Uses the entire training dataset to calculate gradient before updating the weights. 
Stochastic Gradient Descent (SGD): Uses only one training sample at a time to calculate the gradient and update the weights. 
Mini-batch Gradient Descent: Uses a small batch of samples instead of the entire dataset or just one sample. 

The advantage and disadvantage of Batch Gradient Descent, Stochastic Gradient Descent (SGD), and Mini-batch Gradient Descent are given below:
1. Batch Gradient Descent: 
Advantage: Stable and accurate updates.
Disadvantage: Slow and requires more memory for large datasets.
2. Stochastic Gradient Descent (SGD):
Advantage: Faster updates and uses less memory.
Disadvantage: Updates can be noisy and less stable.
3. Mini-batch Gradient Descent:
Advantage: Good balance between speed, memory, and stable updates.
Disadvantage: We need to choose a suitable batch size.

I would recommend Mini-batch Gradient Descent for Fashion-MNIST.
Fashion-MNIST has 60,000 images, so using all images at once can require more memory and be slower. Using only one image at a time can make training unstable.
Mini-batch uses a small number of images, such as 32, 64, or 128, for each update. So, it provides a good balance between speed, memory usage, and stable training.

**Part C Answer:**
1. Fashion-MNIST has 10 mutually exclusive classes, so it is a multiclass classification problem. nn.BCELoss is mainly used for binary classification and nn.CrossEntropyLoss is designed for multiclass classification. Thats why, we use nn.CrossEntropyLoss instead of nn.BCELoss.

2. We should not use Softmax before CrossEntropyLoss because CrossEntropyLoss already handles it internally.

3. loss.backward() calculates the gradients of the loss for all trainable parameters during backpropagation.

2.1 Download the Dataset (1 Marks)
We will use torchvision.datasets.FashionMNIST to download the dataset. This is similar to how we loaded data from .npy files in class, but here PyTorch handles the download for us.

Run the cell below to download the training and test sets:

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import torchvision
import numpy as np
import matplotlib.pyplot as plt

# Class names for Fashion-MNIST
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

# Download Fashion-MNIST — this downloads the data into a ./data folder
# download=True means it will download if not already present
full_train_dataset = torchvision.datasets.FashionMNIST(root='./data', train=True, download=True)
test_dataset_raw = torchvision.datasets.FashionMNIST(root='./data', train=False, download=True)

train_images_all = full_train_dataset.data.numpy()  # shape: (60000, 28, 28)
train_labels_all = full_train_dataset.targets.numpy()  # shape: (60000,)
test_images_all = test_dataset_raw.data.numpy()
test_labels_all = test_dataset_raw.targets.numpy()

print(f"Training set size: {len(full_train_dataset)}")
print(f"Test set size: {len(test_dataset_raw)}")
print(f"Image shape: {full_train_dataset[0][0].size}")
```

2.4 Prepare Tensors and Split (4 Marks)
Now we need to extract the data as NumPy arrays, normalize it, split it, and convert to PyTorch tensors.

Your tasks:

Normalize pixel values to [0, 1] by dividing by 255.0.
Split training data into train (50,000) and validation (10,000) using train_test_split with random_state=42.
Convert all sets to PyTorch tensors — images as float32, labels as long.
Required variables: X_train, X_val, X_test (tensors), y_train, y_val, y_test (tensors)

```python
from sklearn.model_selection import train_test_split

# YOUR CODE HERE

# 1. Normalize pixel values to [0, 1]
X = train_images_all / 255.0
X_test = test_images_all / 255.0

y = train_labels_all
y_test = test_labels_all

# 2. Split training data into 50,000 train and 10,000 validation
X_train, X_val, y_train, y_val = train_test_split(
    X, y,
    test_size=10000,
    random_state=42
)

# 3. Convert to PyTorch tensors
X_train = torch.tensor(X_train, dtype=torch.float32)
X_val = torch.tensor(X_val, dtype=torch.float32)
X_test = torch.tensor(X_test, dtype=torch.float32)

y_train = torch.tensor(y_train, dtype=torch.long)
y_val = torch.tensor(y_val, dtype=torch.long)
y_test = torch.tensor(y_test, dtype=torch.long)

# Check shapes
print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_val:", y_val.shape)
print("y_test:", y_test.shape)
```

```python
# Create TensorDataset objects
train_dataset = TensorDataset(X_train, y_train)
val_dataset = TensorDataset(X_val, y_val)
test_dataset = TensorDataset(X_test, y_test)

# Batch size
batch_size = 64

# Create DataLoaders
train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=batch_size,
    shuffle=False
)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=False
)

# Check number of batches
print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

Question 4: Define the Model (15 Marks)
Create a class FashionClassifier inheriting from nn.Module.

Use nn.Sequential with this architecture:

Layer	Description
Flatten	Convert 28×28 → 784
Linear	784 → 256
ReLU	Activation
Linear	256 → 128
ReLU	Activation
Linear	128 → 64
ReLU	Activation
Linear	64 → 10
Note: Do NOT apply Softmax at the output layer.

After defining the class:

Create an instance of FashionClassifier.
Print the model architecture.

```python
# YOUR CODE HERE
class FashionClassifier(nn.Module):
    def __init__(self):
        super().__init__()

        self.model = nn.Sequential(
            nn.Flatten(),          # 28×28 → 784
            nn.Linear(784, 256),
            nn.ReLU(),

            nn.Linear(256, 128),
            nn.ReLU(),

            nn.Linear(128, 64),
            nn.ReLU(),

            nn.Linear(64, 10)      # 10 classes
        )

    def forward(self, x):
        return self.model(x)


# Create model instance
model = FashionClassifier()

# Print model architecture
print(model)
```

5.1 Set Up Training (5 Marks)

Optimizer: optim.SGD with lr=0.01
Number of epochs: 15

```python
# YOUR CODE HERE
# Loss function
criterion = nn.CrossEntropyLoss()

# Optimizer
optimizer = optim.SGD(model.parameters(), lr=0.01)

# Number of epochs
num_epochs = 15
```

5.2 Training Loop (15 Marks)
Write a complete training loop for 15 epochs.

For each epoch:

Train on train_loader and compute training loss
Evaluate on val_loader (with torch.no_grad()) and compute validation loss
Print: Epoch {epoch+1}/{epochs} | Train Loss: {:.4f} | Val Loss: {:.4f}
Store: train_losses, val_losses, (lists, one value per epoch).

```python
# YOUR CODE HERE
train_losses = []
val_losses = []

train_accuracies = []
val_accuracies = []

epochs = 15

for epoch in range(epochs):

    # Training
    model.train()

    correct_train = 0  
    total_train = 0

    running_train_loss = 0.0

    for images, labels in train_loader:
        # Forward pass
        outputs = model(images)

        # Calculate loss
        loss = criterion(outputs, labels)

        # Clear previous gradients
        optimizer.zero_grad()

        # Backpropagation
        loss.backward()

        # Update weights
        optimizer.step()

        running_train_loss += loss.item()

        # Training accuracy 
        predicted = torch.argmax(outputs, dim=1)
        total_train += labels.size(0)
        correct_train += (predicted == labels).sum().item()

    # Average training loss
    train_loss = running_train_loss / len(train_loader)
    train_losses.append(train_loss)

    # Accuracy
    train_acc = 100.0 * correct_train / total_train
    train_accuracies.append(train_acc)

    # Validation
    model.eval()
    correct_val = 0     
    total_val = 0

    running_val_loss = 0.0

    with torch.no_grad():
        for images, labels in val_loader:

            # Forward pass
            outputs = model(images)

            # validation loss
            loss = criterion(outputs, labels)

            running_val_loss += loss.item()

            # validation accuracy
            predicted = torch.argmax(outputs, dim=1)
            total_val += labels.size(0)
            correct_val += (predicted == labels).sum().item()

    # Average validation loss
    val_loss = running_val_loss / len(val_loader)
    val_losses.append(val_loss)

    # Validation accuracy
    val_acc = 100.0 * correct_val / total_val
    val_accuracies.append(val_acc)

    # Print results
    print(
        f"Epoch {epoch+1}/{epochs} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_loss:.4f}"
    )
```

```python
# --- TEST CELL (DO NOT MODIFY) ---
assert len(train_losses) == 15, f"Expected 15 epochs, got {len(train_losses)}"
assert len(val_losses) == 15
assert train_losses[-1] < train_losses[0], "Training loss should decrease"
print("✅ Q5.2 Passed!")
print(f"   Final Train Acc: {train_accuracies[-1]:.2f}%, Final Val Acc: {val_accuracies[-1]:.2f}%")
```

5.3 Test Evaluation (5 Marks)
Evaluate the model on the test set. Print the test_accuracy (required_variable)

Also, collect all test predictions and true labels into lists called all_preds and all_labels (we'll use them for analysis later).

Print: Test Accuracy: {:.2f}%

```python
# YOUR CODE HERE
model.eval()

all_preds = []
all_labels = []

correct = 0
total = 0

with torch.no_grad():
    for images, labels in test_loader:

        outputs = model(images)
        max_values, predicted = torch.max(outputs, 1)
        all_preds.extend(predicted.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

test_accuracy = 100 * correct / total

print(f"Test Accuracy: {test_accuracy:.2f}%")
```
