```python
import numpy as np
import pandas as pd

np.random.seed(24)

num_employees = 50  # you can increase this if needed

# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)

# Fixed set of roles (repetition allowed)
roles_list = [
    "HR Manager",
    "Software Developer",
    "Accountant",
    "Marketing Executive",
    "Sales Executive",
    "Backend Developer",
    "HR Executive",
    "Financial Analyst",
    "Digital Marketer",
    "Operations Manager"
]

emp_roles = np.random.choice(roles_list, num_employees)

# Create DataFrame
data = pd.DataFrame({
    "emp_id": emp_ids,
    "emp_age": emp_ages,
    "emp_salary": emp_income,
    "role": emp_roles
})

# Save dataset
data.to_csv("staff_data.csv", index=False)
```

```python
# Problem - 1:

import numpy as np

age_arr = np.array(data["emp_age"])
salary_arr = np.array(data["emp_salary"])

print(age_arr)
print(salary_arr)

# Age array info:
print("Data Type:", age_arr.dtype)
print("Number of dimensions:", age_arr.ndim)
print("Shape:", age_arr.shape)
print("Total elements:", age_arr.size)

# Salary array info:
print("Data Type:", salary_arr.dtype)
print("Number of dimensions:", salary_arr.ndim)
print("Shape:", salary_arr.shape)
print("Total elements:", salary_arr.size)
```

```python
# Problem - 2:

max_salary = np.max(salary_arr)
print(max_salary)

min_salary = np.min(salary_arr)
print(min_salary)

avg_salary = np.mean(salary_arr)
print(avg_salary)

avg_age = np.mean(age_arr)
print(avg_age)
```

```python
# Problem - 3:

all_conditions = (age_arr >= 25) & (age_arr <= 40) & (salary_arr > 50000)

filter_age = age_arr[all_conditions]
print(filter_age)

filter_salary = salary_arr[all_conditions]
print(filter_salary)

total_employees_count = np.sum(all_conditions)
print(total_employees_count)
```

```python
# Problem - 4:

new_age = np.where(
    age_arr < avg_age,
    age_arr + 4,
    age_arr + 2
)

print("Original array:", age_arr)
print("New_age:", new_age)
print("Average age:", avg_age)
```

```python
# Problem - 5:

salary = data["emp_salary"].values
tech_mask = data["role"].isin(roles_list).values
new_salaries = salary.copy()

new_salaries[tech_mask] = new_salaries[tech_mask] * 1.08

new_total = np.sum(new_salaries)
print("New Total Salary Expense:", new_total)

old_total = np.sum(salary)
print("Old Total Salary Expense:", old_total)

difference = new_total - old_total
print("Increase in Expense:", difference)
```

```python
# Problem - 10:

df["emp_salary"] = df["emp_salary"].astype(float)

df["prev_salary"] = df["emp_salary"]
tech_role = ["Software Developer", "Backend Developer"]

df.loc[df["role"].isin(tech_role), "emp_salary"] *= 1.12
df.loc[~df["role"].isin(tech_role), "emp_salary"]  *= 1.05

df["new_salary"] = df["emp_salary"]
df["increase_salary"] = df["new_salary"] - df["prev_salary"]

final_ans = df[["role", "prev_salary", "new_salary", "increase_salary"]]
print(final_ans)
```
