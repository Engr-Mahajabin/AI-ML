```python
import pandas as pd
import matplotlib.pyplot as plt

# ১. আপনার দেওয়া ডেটা দিয়ে ডিকশনারি তৈরি
trend_data = {
    "Year": [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024],
    "Programming": [589, 862, 1085, 1150, 1439, 1715, 1785, 1995, 2236, 2442],
    "Digital Marketing": [734, 963, 1157, 1404, 1497, 1579, 1948, 2073, 2310, 2338]
}

# ২. পান্ডাস ডেটাফ্রেম তৈরি
df = pd.DataFrame(trend_data)

# ৩. সিএসভি (CSV) ফাইল হিসেবে সেভ করা
df.to_csv("enrollment_data.csv", index=False)
print("ডেটাফ্রেম তৈরি হয়েছে এবং 'enrollment_data.csv' নামে সেভ হয়েছে!\n")
print(df)

print("\n" + "="*50 + "\n")

# ৪. Matplotlib দিয়ে একটি সুন্দর গ্রাফ বা প্লট তৈরি করা
plt.figure(figsize=(10, 6))

# Programming-এর জন্য লাইন
plt.plot(df["Year"], df["Programming"], marker='o', linewidth=2, color='blue', label='Programming')

# Digital Marketing-এর জন্য লাইন
plt.plot(df["Year"], df["Digital Marketing"], marker='s', linewidth=2, color='green', label='Digital Marketing')

# গ্রাফের টাইটেল ও লেবেল সেট করা
plt.title("Growth Trend: Programming vs Digital Marketing (2015-2024)", fontsize=14, fontweight='bold')
plt.xlabel("Year", fontsize=12)
plt.ylabel("Value/Enrollment", fontsize=12)

# গ্রিড এবং লেজেন্ড (ইনডিকেটর) দেখানো
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)

# বছরের টিকসগুলো (X-axis) সুন্দরভাবে দেখানোর জন্য
plt.xticks(df["Year"])

# গ্রাফটি স্ক্রিনে শো করা
plt.show()
```

```python
plt.xlabel('Years') 
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')

plt.plot(df['Year'] ,df['Programming']) 
```

```python
#  Bar chart on real data 

df = pd.read_csv('student_data.csv') 

df
```
