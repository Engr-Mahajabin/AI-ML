```python
# You must run this code and use this dataset for all questions
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
num_students = 50

# Student data
student_ids = np.arange(201, 201 + num_students)
student_ages = np.random.randint(15, 22, num_students)
student_scores = np.random.randint(40, 100, num_students)
student_attendance = np.random.randint(60, 100, num_students)  # percentage

departments_list = [
    "Science",
    "Commerce",
    "Arts",
    "Computer Science",
    "Mathematics",
    "English",
    "Physics",
    "Chemistry",
    "Biology",
    "Economics"
]

student_departments = np.random.choice(departments_list, num_students)

grade_conditions = [
    student_scores >= 80,
    (student_scores >= 65) & (student_scores < 80),
    (student_scores >= 50) & (student_scores < 65),
    student_scores < 50
]
grade_values = ["A", "B", "C", "F"]
student_grades = np.select(grade_conditions, grade_values, default="F")

# Create DataFrame
data = pd.DataFrame({
    "student_id": student_ids,
    "student_age": student_ages,
    "score": student_scores,
    "attendance": student_attendance,
    "department": student_departments,
    "grade": student_grades
})

# Save dataset
data.to_csv("school_data.csv", index=False)
```

```python
#Problem -4:

data["prev_score"] = data["score"]

data.loc[data["department"].isin(["Computer Science", "Mathematics"]),"score"] *= 1.10

data.loc[~data["department"].isin(["Computer Science", "Mathematics"]),"score"] *= 1.05

data["score"] = data["score"].clip(upper=100)

data["new_score"] = data["score"]

data["increase_score"] = (data["new_score"] - data["prev_score"])

print(data[["student_id", "department", "prev_score","new_score","increase_score"]])
```

```python
# Problem -5:

import matplotlib.pyplot as plt

avg_scores = data.groupby('department')['score'].mean()

plt.bar(avg_scores.index, avg_scores.values, color='gray', edgecolor='black')

plt.xlabel('Department names')
plt.ylabel('Average score')
plt.title('Average Score by Department')

plt.xticks(rotation=45)

plt.show()
```
